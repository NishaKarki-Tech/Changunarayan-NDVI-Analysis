```python
import pandas as pd

df = pd.read_csv("Changunarayan_NDVI_2025_Final.csv")

df
```

```python
import matplotlib.pyplot as plt

# Month names
month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

# Add month names
df["Month_Name"] = month_names

# Create figure
plt.figure(figsize=(9,5))

plt.plot(
    df["Month_Name"],
    df["Mean_NDVI"],
    marker="o",
    linewidth=2
)

plt.title("Monthly NDVI Variation in Changunarayan Municipality (2025)", fontsize=14)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Mean NDVI", fontsize=12)

plt.ylim(0.3, 0.8)

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(9,5))

plt.plot(
    df["Month_Name"],
    df["Mean_NDVI"],
    marker="o",
    linewidth=2
)

plt.title("Monthly NDVI Variation in Changunarayan Municipality (2025)", fontsize=14)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Mean NDVI", fontsize=12)

plt.ylim(0.3, 0.8)

plt.grid(True)

plt.savefig(
    "Monthly_NDVI_Trend_2025.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Assign seasons
season_map = {
    1: "Winter",
    2: "Winter",
    3: "Pre-monsoon",
    4: "Pre-monsoon",
    5: "Pre-monsoon",
    6: "Monsoon",
    7: "Monsoon",
    8: "Monsoon",
    9: "Monsoon",
    10: "Post-monsoon",
    11: "Post-monsoon",
    12: "Post-monsoon"
}

df["Season"] = df["Month"].map(season_map)

seasonal_mean = (
    df.groupby("Season")["Mean_NDVI"]
      .mean()
      .reset_index()
)

print(seasonal_mean)
```

```python
import matplotlib.pyplot as plt

# Order the seasons
season_order = [
    "Winter",
    "Pre-monsoon",
    "Monsoon",
    "Post-monsoon"
]

seasonal_mean = seasonal_mean.set_index("Season").loc[season_order].reset_index()

plt.figure(figsize=(8,5))

plt.bar(
    seasonal_mean["Season"],
    seasonal_mean["Mean_NDVI"],
    edgecolor="black"
)

plt.title("Seasonal Mean NDVI in Changunarayan Municipality (2025)")
plt.xlabel("Season")
plt.ylabel("Mean NDVI")

plt.ylim(0.3, 0.7)

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    seasonal_mean["Season"],
    seasonal_mean["Mean_NDVI"],
    edgecolor="black"
)

plt.title("Seasonal Mean NDVI in Changunarayan Municipality (2025)")
plt.xlabel("Season")
plt.ylabel("Mean NDVI")

plt.ylim(0.3, 0.7)

plt.grid(axis="y")

plt.savefig(
    "Seasonal_NDVI_Comparison_2025.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
